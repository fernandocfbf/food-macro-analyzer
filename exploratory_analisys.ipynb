# FoodSeg103 Dataset Exploration

This notebook aims to explore the **FoodSeg103** dataset, which contains images of various food items along with their segmentation masks. The goal is to analyze the dataset structure, visualize sample images, and prepare the data for further processing in computer vision tasks, such as image segmentation and classification.
